```python
import numpy as np
import matplotlib.pyplot as plt
from collections import defaultdict

# Função para calcular métricas e plotar gráficos
def plot_metrics_by_bc_ech(data):
    # Dicionário para armazenar as métricas por BC e Ech
    metrics = defaultdict(lambda: defaultdict(lambda: {
        'infractions': [],
        'distance_before_infraction': [],
        'left_trajectories': [],
        'right_trajectories': []
    }))

    # Percorrer os dados e calcular as métricas
    for entry in data:
        BC, M, Ech, Ep, values = entry
        x = values[0]  # x está na primeira coluna
        y = values[1]  # y está na segunda coluna
        
        # Variáveis para verificar infrações
        stop_infraction = values[24]
        encounter_stop = values[23]
        run_red_light = values[22]
        outside_lane = values[21]
        wrong_lane = values[20]
        collisions_layout = values[15]
        collisions_vehicle = values[16]
        collisions_pedestrian = values[17]
        route_completed_in_m = values[3]
        c_blocked = values[6]
        c_lat_dist = values[7]
        c_collision = values[8]
        collision = values[9]
        c_collision_px = values[10]
        timeout = values[11]
        
        # Calculando a quantidade de infrações
        infractions = 1 if sum(c_blocked) + sum(c_lat_dist) + sum(collisions_layout) + sum(timeout) + sum(outside_lane) + sum(wrong_lane) else 0
        metrics[BC][Ech]['infractions'].append(infractions)
        
        # Calculando a distância percorrida antes da primeira infração
        distance_before_infraction = route_completed_in_m[-1]
        for i in range(1, len(x)):
            if c_blocked[i] == 1 or c_lat_dist[i] == 1 or collisions_layout[i] == 1 or timeout[i] == 1 or outside_lane[i] == 1 or wrong_lane[i] == 1:
                distance_before_infraction = route_completed_in_m[i]
                break
        metrics[BC][Ech]['distance_before_infraction'].append(distance_before_infraction)
        
        # Calculando a quantidade de trajetórias à esquerda
        last_x = x[-1]
        if last_x < -0.0018:
            metrics[BC][Ech]['left_trajectories'].append(1)
        else:
            metrics[BC][Ech]['left_trajectories'].append(0)

        # Calculando a quantidade de trajetórias à direita
        if last_x > -0.0015:
            metrics[BC][Ech]['right_trajectories'].append(1)
        else:
            metrics[BC][Ech]['right_trajectories'].append(0)
    
    # Agora plotamos a evolução das métricas ao longo de Ech para cada BC
    for BC in sorted(metrics.keys()):
        ech_values = sorted(metrics[BC].keys())
        
        # Preparar listas para armazenar a média e o desvio padrão das métricas
        infractions_sum = []
        infractions_std = []
        distance_mean = []
        distance_std = []
        left_sum = []
        left_std = []
        right_sum = []
        right_std = []

        # Calcular a média e o desvio padrão para cada Ech
        for Ech in ech_values:
            infractions_sum.append(np.sum(metrics[BC][Ech]['infractions']))
            infractions_std.append(np.std(metrics[BC][Ech]['infractions']))
            
            distance_mean.append(np.mean(metrics[BC][Ech]['distance_before_infraction']))
            distance_std.append(np.std(metrics[BC][Ech]['distance_before_infraction']))
            
            left_sum.append(np.sum(metrics[BC][Ech]['left_trajectories']))
            left_std.append(np.std(metrics[BC][Ech]['left_trajectories']))
            
            right_sum.append(np.sum(metrics[BC][Ech]['right_trajectories']))
            right_std.append(np.std(metrics[BC][Ech]['right_trajectories']))

        # Plotar infrações
        plt.figure(figsize=(8, 6))
        plt.errorbar(ech_values, infractions_sum, yerr=infractions_std, fmt='-o', label='Infractions', color='red')
        plt.fill_between(ech_values, np.array(infractions_sum) - np.array(infractions_std), np.array(infractions_sum) + np.array(infractions_std), color='red', alpha=0.2)
        plt.title(f', for BC={BC}')
        plt.xlabel('Ech')
        plt.ylabel('Infractions')
        plt.grid(True)
        plt.show()

        # Plotar distância antes da infração
        plt.figure(figsize=(8, 6))
        plt.errorbar(ech_values, distance_mean, yerr=distance_std, fmt='-o', label='Distance Before Infraction', color='blue')
        plt.fill_between(ech_values, np.array(distance_mean) - np.array(distance_std), np.array(distance_mean) + np.array(distance_std), color='blue', alpha=0.2)
        plt.title(f'Distance Before Infraction Evolution for BC={BC}')
        plt.xlabel('Ech')
        plt.ylabel('Distance (m)')
        plt.grid(True)
        plt.show()

        # Plotar trajetórias à esquerda
        plt.figure(figsize=(8, 6))
        plt.errorbar(ech_values, left_sum, yerr=left_std, fmt='-o', label='Left Trajectories', color='green')
        plt.fill_between(ech_values, np.array(left_sum) - np.array(left_std), np.array(left_sum) + np.array(left_std), color='green', alpha=0.2)
        plt.title(f'Left Trajectories Evolution for BC={BC}')
        plt.xlabel('Ech')
        plt.ylabel('Left Trajectories')
        plt.grid(True)
        plt.show()

        # Plotar trajetórias à direita
        plt.figure(figsize=(8, 6))
        plt.errorbar(ech_values, right_sum, yerr=right_std, fmt='-o', label='Right Trajectories', color='orange')
        plt.fill_between(ech_values, np.array(right_sum) - np.array(right_std), np.array(right_sum) + np.array(right_std), color='orange', alpha=0.2)
        plt.title(f'Right Trajectories Evolution for BC={BC}')
        plt.xlabel('Ech')
        plt.ylabel('Right Trajectories')
        plt.grid(True)
        plt.show()

# Caminho para os dados e chamada da função
path_to_data_folder_2 = "diff_bc_video_(diffuser)/birdview/town01_multimodality_t_intersection_simples_GKC"
data = np.load(path_to_data_folder_2 + '/data.npy', allow_pickle=True)

plot_metrics_by_bc_ech(data)
```
